```python
# convertinf transaction_date to datetime format
df["transaction_date"]=pd.to_datetime(df["transaction_date"],errors="coerce")
```

```python
df["transaction_datetime"] = pd.to_datetime(
    df["transaction_date"].dt.strftime("%Y-%m-%d") + " " + df["transaction_time"],
    format="%Y-%m-%d %H:%M",
    errors="coerce"
)
```

```python
df[["transaction_date", "transaction_time", "transaction_datetime"]].head()
```

```python
df["transaction_datetime"].isnull().sum()
```

```python
df["year"]=df["transaction_datetime"].dt.year
df["month"]=df["transaction_datetime"].dt.month
df["month_name"]=df["transaction_datetime"].dt.month_name()
df["day_of_week"]=df["transaction_datetime"].dt.day_name()
df["is_weekend"]=df["transaction_datetime"].dt.day_of_week>=5
```

```python
df[
    ["transaction_datetime","year","month","month_name","day_of_week","is_weekend","transaction_hour"]
].head(10)
     
```

### EDA

```python
df.groupby("year").agg(
    transactions=("transaction_id","count"),
    total_amount=("transaction_amount","sum"),
    avg_amount=("transaction_amount","mean"),
).sort_values("total_amount", ascending=False).round(2)    # round= reduces the number of decimal places to 2                   
```

```python
df.groupby("is_fraud").agg(
    transactions=("transaction_id","count"),
        total_amount=("transaction_amount","sum"),
        avg_amount=("transaction_amount","mean"),
).round(2)
```

```python
df[df["is_fraud"]==1].groupby("transaction_type")["transaction_amount"].agg(["count","sum","mean"]).round(2).sort_values("sum",ascending=False)
```

```python
df.groupby("customer_id").agg(
    transactions=("transaction_id","count"),
    total_amount=("transaction_amount","sum"),
    avg_amount=("transaction_amount","mean")
).sort_values("total_amount",ascending=False).head(10).round(2)
```

```python
merchant_analysis = df.groupby("merchant_category").agg(
    transactions=("transaction_id", "count"),
    total_amount=("transaction_amount", "sum"),
    avg_amount=("transaction_amount", "mean")
).sort_values("total_amount", ascending=False).round(2)

merchant_analysis
```

### BUSINESS KPIs

```python
kpis={
    "total transactions":df["transaction_id"].nunique(),
    "total transaction value":df["transaction_amount"].sum(),
    "average transaction value":df["transaction_amount"].mean(),
    "total customers":df["customer_id"].nunique(),
    "fraud transactions":df["is_fraud"].sum(),
    "fraud rate":df["is_fraud"].mean()*100

}

pd.Series(kpis).round(2)
```
